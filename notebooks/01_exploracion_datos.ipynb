**Objetivo:** explorar el dataset, entender su estructura y detectar problemas de calidad antes de limpiarlo y analizarlo.

**Fuente:** Kaggle, *Amazon Top 50 Bestselling Books 2009-2019*.

```python

# Cargar todas las librerías
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from scipy import stats as st
import math as mt

# Muestra todas las columnas al imprimir tablas
pd.set_option("display.max_columns", None)
```

```python
# CARGAMOS LOS DATOS
ruta = "../data/amazon_bestsellers_2009-2019.csv"
df = pd.read_csv(ruta)
```

```python
# Información general/resumida sobre el DataFrame
print('Numero de filas y columnas de la tabla ruta: ', df.shape)
print('Informacion general de la tabla df: ')
print()
df.info()
print()
print()
# Muestra de los datos
print('las primeras 50 filas de la tabla df: ')
print(df.head(51))
print()
print()
# observar si tiene duplicados
print('La tabla df tiene ', df.duplicated().sum(), ' filas duplicadas')
```

```python
# Análisis detallado de valores ausentes
print("=== ANÁLISIS DE VALORES AUSENTES ===")
missing_data = df.isnull().sum()
missing_percentage = (missing_data / len(df)) * 100

for column in df.columns:
    if missing_data[column] > 0:
        print(f"{column}: {missing_data[column]} ausentes ({missing_percentage[column]:.1f}%)")
print()
print()
# Valores únicos de columnas categóricas
print("Géneros:", df["Genre"].unique())
print("Años:", sorted(df["Year"].unique()))
print("Libros únicos:", df["Name"].nunique())
print("Autores únicos:", df["Author"].nunique())
print()
print()
# Estadísticas descriptivas
print("\n=== ESTADÍSTICAS DESCRIPTIVAS ===")
print(df.describe())
```
